Rasporedjivanje slika po direktorijumima za upload na label studio

```python
from pathlib import Path
import shutil

src_root = Path.cwd().parents[1] / " " # put do foldera za rad
if not src_root.exists():
    raise FileNotFoundError(f"{src_root} ne postoji")

target_root = Path.cwd() / " " # naziv ciljnog direktorijuma za rad
target_root.mkdir(parents=True, exist_ok=True)

MAX_PER_FOLDER = 100

count = 0
folder_index = 1
file_in_folder = 0

current_target = target_root / f"batch_{folder_index:03d}"
current_target.mkdir(exist_ok=True)

for p in src_root.rglob("*"):
    if p.is_file() and p.suffix.lower() == ".jpg":
        if file_in_folder >= MAX_PER_FOLDER:
            folder_index += 1
            file_in_folder = 0
            current_target = target_root / f"batch_{folder_index:03d}"
            current_target.mkdir(exist_ok=True)

        dest = current_target / p.name

        if dest.exists():
            stem, suf = dest.stem, dest.suffix
            i = 1
            while True:
                new_name = f"{stem}_{i}{suf}"
                new_dest = current_target / new_name
                if not new_dest.exists():
                    dest = new_dest
                    break
                i += 1

        shutil.copy2(p, dest)
        count += 1
        file_in_folder += 1

print(f"Kopirano je {count} .jpg fajlova u {folder_index} foldera unutar {target_root}")
```
